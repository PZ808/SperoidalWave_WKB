# Tilted action-frame versus compact Griffiths--Dwork Gauss--Manin system

This notebook generalizes the symmetric $s=0$ harmonization calculation to the tilted, spin-dependent case.  The goal is to compare two different Gauss--Manin descriptions of the same genus-one family:

1. a meromorphic action-frame system built from the action differential and its parameter derivatives, and

2. a compact Griffiths--Dwork-style system built from a pole-cleaned two-dimensional cohomology frame.

The main point is that these two computations are not written in the same frame.  The action-frame system naturally sees meromorphic differentials with poles at $x=\pm 1$, while the compact Griffiths--Dwork system works with pole-cleaned representatives.  The notebook computes both systems and checks that they agree after the correct descent from the meromorphic frame to the compact frame.

---

## 1. Tilted affine curve

We work with the affine form of the tilted curve

$$
\frac12 y^2
+x^4
-\mu_2 x^3
+(E-2)x^2
+(\mu_1+\mu_2)x
+\mu_0-E+1
=0.
$$

Equivalently,

$$
y^2=P(x),
$$

where

$$
P(x)
=
-2\left[
x^4
-\mu_2x^3
+(E-2)x^2
+(\mu_1+\mu_2)x
+\mu_0-E+1
\right].
$$

We also define the finite-pole divisor

$$
D(x)=1-x^2.
$$

All one-forms are represented in the form

$$
r(x)\frac{dx}{y}.
$$

---

## 2. Exact differentials

Exact differentials are generated by expressions of the form

$$
d(h(x)y).
$$

Using $y^2=P(x)$, we have

$$
d(hy)
=
\left(
P h'
+
\frac12 P'h
\right)\frac{dx}{y}.
$$

Therefore two coefficients $r_1(x)$ and $r_2(x)$ represent the same cohomology class if

$$
r_1-r_2
=
L(h),
$$

where

$$
\boxed{
L(h)
=
P h'
+
\frac12 P'h.
}
$$

This is the exact-reduction operator used throughout the notebook.

---

## 3. Parameter derivatives

For a differential

$$
\omega
=
r(x)\frac{dx}{y},
$$

the parameter derivative is

$$
\partial_\lambda \omega
=
\left[
\partial_\lambda r
-
\frac{r\,\partial_\lambda P}{2P}
\right]\frac{dx}{y}.
$$

For the tilted polynomial $P(x)$, the relevant parameter derivatives are

$$
\partial_E P
=
2(1-x^2),
$$

$$
\partial_{\mu_0}P
=
-2,
$$

$$
\partial_{\mu_1}P
=
-2x,
$$

and

$$
\partial_{\mu_2}P
=
2x^3-2x
=
-2x(1-x^2).
$$

---

## 4. Meromorphic tilted action frame

The classical action differential is

$$
\omega_0
=
\frac{y\,dx}{1-x^2}
=
\frac{P}{D}\frac{dx}{y}.
$$

Taking parameter derivatives of $\omega_0$ gives the tilted action-frame basis

$$
\boxed{
\begin{aligned}
\omega_0
&=
\frac{P}{D}\frac{dx}{y},
\\
\omega_E
&=
\frac{dx}{y},
\\
\omega_{\mu_0}
&=
-\frac1D\frac{dx}{y},
\\
\omega_{\mu_1}
&=
-\frac{x}{D}\frac{dx}{y},
\\
\omega_{\mu_2}
&=
-x\frac{dx}{y}.
\end{aligned}
}
$$

Thus the meromorphic action frame is

$$
\boxed{
\omega
=
\begin{pmatrix}
\omega_0\\
\omega_E\\
\omega_{\mu_0}\\
\omega_{\mu_1}\\
\omega_{\mu_2}
\end{pmatrix}.
}
$$

The notebook computes the meromorphic Gauss--Manin matrices $U_\lambda$ defined by the column-image convention

$$
\boxed{
\partial_\lambda \omega_j
=
\sum_i (U_\lambda)_{ij}\omega_i
\quad
\text{mod exact forms}.
}
$$

Equivalently,

$$
\partial_\lambda \omega
=
U_\lambda \omega.
$$

Here

$$
\lambda\in\{E,\mu_0,\mu_1,\mu_2\}.
$$

---

## 5. Pole-cleaned compact frame

The action frame contains meromorphic representatives.  To compare with the compact Griffiths--Dwork reduction, we pass to a pole-cleaned compact frame.

We define

$$
\eta_0
=
\omega_E
=
\frac{dx}{y},
$$

and

$$
\eta_1
=
\omega_0
-2\mu_0\omega_{\mu_0}
-2\mu_1\omega_{\mu_1}
-\mu_2\omega_{\mu_2}.
$$

Using the explicit action-frame representatives, this becomes

$$
\eta_1
=
\left[
\frac{P}{D}
+\frac{2\mu_0}{D}
+\frac{2\mu_1x}{D}
+\mu_2x
\right]\frac{dx}{y}.
$$

Putting everything over the common denominator $D$, one finds

$$
P+2\mu_0+2\mu_1x+\mu_2xD
=
D\left[2(x^2+E-1)-\mu_2x\right].
$$

Therefore

$$
\boxed{
\eta_1
=
\left[2(x^2+E-1)-\mu_2x\right]\frac{dx}{y}.
}
$$

So the compact frame is

$$
\boxed{
\eta
=
\begin{pmatrix}
\eta_0\\
\eta_1
\end{pmatrix}
=
\begin{pmatrix}
1\\
2(x^2+E-1) -\mu_2 x
\end{pmatrix}
\frac{dx}{y}.
}
$$

Compared with the symmetric calculation, the compact representative has an extra $-\mu_2x$ term.  This is the residue of the cubic tilt: the $\mu_2$ contribution is not pole-cleaned with a factor of $2$, unlike the finite-pole terms proportional to $\mu_0$ and $\mu_1$.  All reductions are then performed using the tilted polynomial $P(x)$.

---

## 6. Embedding of the compact frame into the meromorphic frame

The compact frame embeds into the meromorphic action frame by

$$
\eta_0
=
\omega_E,
$$

and

$$
\eta_1
=
\omega_0
-2\mu_0\omega_{\mu_0}
-2\mu_1\omega_{\mu_1}
-\mu_2\omega_{\mu_2}.
$$

Thus, with the action-frame ordering

$$
\omega
=
\begin{pmatrix}
\omega_0\\
\omega_E\\
\omega_{\mu_0}\\
\omega_{\mu_1}\\
\omega_{\mu_2}
\end{pmatrix},
$$

the embedding matrix is

$$
\boxed{
S
=
\begin{pmatrix}
0 & 1\\
1 & 0\\
0 & -2\mu_0\\
0 & -2\mu_1\\
0 & -\mu_2
\end{pmatrix}.
}
$$

The columns of $S$ are the action-frame coordinates of $\eta_0$ and $\eta_1$.

---

## 7. Descent equation

Because the $U_\lambda$ matrices use the column-image convention, the compact connection $B_\lambda$ is obtained from the descent equation

$$
\boxed{
\partial_\lambda S
+
U_\lambda S
=
S B_\lambda
\quad
\text{mod exact forms}.
}
$$

This equation should not be imposed as a raw equality of unreduced rational representatives.  Instead, for each column $j$, the notebook computes

$$
v_j
=
\partial_\lambda S_{\cdot j}
+
U_\lambda S_{\cdot j},
$$

interprets $v_j$ as a meromorphic action-frame combination

$$
v_j^0\omega_0
+
v_j^E\omega_E
+
v_j^{\mu_0}\omega_{\mu_0}
+
v_j^{\mu_1}\omega_{\mu_1}
+
v_j^{\mu_2}\omega_{\mu_2},
$$

converts this back to a single rational coefficient multiplying $dx/y$, and then reduces it modulo exact forms into the compact basis

$$
\{\eta_0,\eta_1\}.
$$

This produces the compact connection from action-frame descent:

$$
\boxed{
\partial_\lambda\eta
=
B_\lambda^{\rm from\ action}\eta.
}
$$

---

## 8. Direct compact Griffiths--Dwork check

Independently, the notebook computes the compact connection directly from

$$
\eta_0
=
\frac{dx}{y},
$$

and

$$
\eta_1
=
\left[2(x^2+E-1)-\mu_2x\right]\frac{dx}{y}.
$$

For each $\lambda\in\{E,\mu_0,\mu_1,\mu_2\}$, it computes

$$
\partial_\lambda\eta_j
=
\left[
\partial_\lambda r_j
-
\frac{r_j\,\partial_\lambda P}{2P}
\right]\frac{dx}{y},
$$

and reduces the result into the compact basis modulo exacts generated by

$$
L(h)
=
Ph'
+
\frac12P'h.
$$

This produces a direct compact connection

$$
\boxed{
\partial_\lambda\eta
=
B_\lambda^{\rm direct}\eta.
}
$$

---

## 9. Harmonization test

The central check of the notebook is

$$
\boxed{
B_\lambda^{\rm from\ action}
-
B_\lambda^{\rm direct}
=
0,
\qquad
\lambda\in\{E,\mu_0,\mu_1,\mu_2\}.
}
$$

Passing this check means that the meromorphic tilted action-frame Gauss--Manin system correctly descends to the same compact system obtained by direct Griffiths--Dwork reduction.

In other words, the diagram

$$
\boxed{
\begin{array}{ccc}
\text{tilted meromorphic action frame}
&
\xrightarrow{\quad S\quad}
&
\text{pole-cleaned compact frame}
\\[0.4em]
(\omega_0,\omega_E,\omega_{\mu_0},\omega_{\mu_1},\omega_{\mu_2})
&&
(\eta_0,\eta_1)
\\[0.4em]
U_\lambda
&&
B_\lambda
\end{array}
}
$$

commutes after reduction modulo exact forms.

---

## 10. Perturbative interpretation

For the perturbative spin-tilted problem one may introduce a bookkeeping parameter $\epsilon$ by writing

$$
\mu_1=\epsilon\nu_1,
\qquad
\mu_2=\epsilon\nu_2.
$$

Then

$$
P=P^{(0)}+\epsilon P^{(1)},
$$

with

$$
P^{(0)}
=
-2\left[
x^4
+(E-2)x^2
+\mu_0-E+1
\right],
$$

and

$$
P^{(1)}
=
-2\left[
-\nu_2x^3
+
(\nu_1+\nu_2)x
\right].
$$

The exact tilted matrices may then be expanded in $\epsilon$ to obtain the perturbative compact Gauss--Manin system relevant for the spin-dependent WKB problem.

The important structural identity remains

$$
\boxed{
\omega_0
-2\mu_0\omega_{\mu_0}
-2\mu_1\omega_{\mu_1}
-\mu_2\omega_{\mu_2}
=
\left[2(x^2+E-1)-\mu_2x\right]\frac{dx}{y}.
}
$$

This is the tilted analogue of the symmetric pole-cleaning identity, and it is the key reason the action-frame and compact Griffiths--Dwork calculations can be harmonized.

```python
# ============================================================
# Tilted meromorphic action-frame GM system vs compact GD frame
# ============================================================
#
# Curve:
#
#   1/2 y^2 + x^4 - mu2 x^3 + (E-2)x^2
#       + (mu1+mu2)x + mu0 - E + 1 = 0
#
# or
#
#   y^2 = P(x)
#
# with
#
#   P(x) = -2[ x^4 - mu2 x^3 + (E-2)x^2
#              + (mu1+mu2)x + mu0 - E + 1 ].
#
# Differentials are represented as
#
#   r(x) dx/y.
#
# Exact forms are generated by
#
#   d(h y) = (P h' + 1/2 P' h) dx/y.
#
# ============================================================


# ------------------------------------------------------------
# 0.a Rings
# ------------------------------------------------------------

Kbase = PolynomialRing(QQ, ['E', 'mu0', 'mu1', 'mu2'])
E, mu0, mu1, mu2 = Kbase.gens()

K = Kbase.fraction_field()

R = PolynomialRing(K, 'x')
x = R.gen()

Fx = FractionField(R)

# ------------------------------------------------------------
# 0.b Utils
# ------------------------------------------------------------

def factor0(x):
    x = K(x)
    if x == 0:
        return K(0)
    return x.factor()

def latex_factor0(x):
    x = K(x)

    if x == 0:
        return "0"

    num = x.numerator()
    den = x.denominator()

    if den == 1:
        return latex(num.factor())

    return r"\frac{%s}{%s}" % (latex(num.factor()), latex(den.factor()))

def show_factored_matrix(M, name="M"):
    rows = []
    for i in range(M.nrows()):
        row = []
        for j in range(M.ncols()):
            row.append(latex_factor0(M[i,j]))
        rows.append(" & ".join(row))

    body = r"\begin{pmatrix}" + r"\\".join(rows) + r"\end{pmatrix}"
    show(LatexExpr(name + "=" + body))

# ------------------------------------------------------------
# 1. Curve data
# ------------------------------------------------------------

D = Fx(1 - x^2)

P = Fx(-2 * (
    x^4
    - mu2*x^3
    + (E - 2)*x^2
    + (mu1 + mu2)*x
    + mu0 - E + 1
))

Px = P.derivative(x)

params = [E, mu0, mu1, mu2]
param_names = ['E', 'mu0', 'mu1', 'mu2']


def dP_param(lam):
    """
    Parameter derivative of P.
    """
    return Fx(P.derivative(lam))


dP = {
    E:    dP_param(E),
    mu0:  dP_param(mu0),
    mu1:  dP_param(mu1),
    mu2:  dP_param(mu2),
}

print("Parameter derivatives of P:")
for lam in params:
    print(lam, ":", dP[lam])


# ------------------------------------------------------------
# 2. Exact-reduction operator
# ------------------------------------------------------------

def L_exact(h):
    """
    Exact form generator:
        L(h) = P h' + 1/2 P' h.
    """
    h = Fx(h)
    return Fx(P*h.derivative(x) + Fx(1)/2 * Px*h)


# ------------------------------------------------------------
# 3. Differential derivative rule
# ------------------------------------------------------------

def param_deriv_form(r, lam):
    """
    For omega = r(x) dx/y,

        d_lam omega =
        [d_lam r - r d_lam P/(2P)] dx/y.
    """
    r = Fx(r)
    return Fx(r.derivative(lam) - r*dP[lam]/(2*P))


# ------------------------------------------------------------
# 4. Meromorphic tilted action frame
# ------------------------------------------------------------

# omega_0       = P/D dx/y
# omega_E       = 1 dx/y
# omega_mu0     = -1/D dx/y
# omega_mu1     = -x/D dx/y
# omega_mu2     = -x dx/y

omega_basis = [
    Fx(P/D),
    Fx(1),
    Fx(-1/D),
    Fx(-x/D),
    Fx(-x)
]

omega_names = ['omega0', 'omegaE', 'omegamu0', 'omegamu1', 'omegamu2']

n_omega = len(omega_basis)

# ------------------------------------------------------------
# 5. Compact pole-cleaned frame
# ------------------------------------------------------------

# eta_0 = omega_E
#
# eta_1 = omega_0 - 2 mu0 omega_mu0
#                 - 2 mu1 omega_mu1
#                 -   mu2 omega_mu2
#
# This simplifies to
#
# eta_1 = [2(x^2 + E - 1) - mu2*x] dx/y.
#
# Important:
# The mu2 coefficient is -mu2, not -2*mu2.  The -2 factors clean
# the finite-pole residues associated with mu0 and mu1, while the
# cubic tilt mu2 also affects the behavior at infinity.

eta_basis = [
    Fx(1),
    Fx(2*(x^2 + E - 1) - mu2*x)
]

eta_names = ['eta0', 'eta1']
n_eta = len(eta_basis)


# Sanity check for pole cleaning
eta1_from_action = (
    omega_basis[0]
    - 2*mu0*omega_basis[2]
    - 2*mu1*omega_basis[3]
    - mu2*omega_basis[4]
)

eta1_clean = eta_basis[1]

print("\nPole-cleaning check:")
print("eta1_from_action - eta1_clean =")
print(Fx(eta1_from_action - eta1_clean))

assert Fx(eta1_from_action - eta1_clean) == 0

# ------------------------------------------------------------
# 6. Linear reduction machinery
# ------------------------------------------------------------

def coeff_equations(expr):
    """
    Given a rational function expr in x over K, clear denominators and
    return the coefficients of the numerator polynomial.
    """
    expr = Fx(expr)
    num = R(expr.numerator())
    return [num.monomial_coefficient(x^k) for k in range(num.degree() + 1)]


def solve_linear_system(eqs, unknowns):
    """
    Solve a linear system whose equations are polynomials in the unknowns
    with coefficients in K.

    Each equation has the form

        const + a0*u0 + a1*u1 + ... = 0.

    We extract the coefficients a_i and the constant term without trying
    to coerce the whole polynomial into K.
    """
    rows = []
    rhs = []

    for eq in eqs:
        parent = eq.parent()
        eq = parent(eq)

        const = eq.constant_coefficient()

        row = []
        recon = const

        for u in unknowns:
            a = eq.monomial_coefficient(u)
            row.append(K(a))
            recon += a*u

        # Sanity check: make sure no nonlinear unknown terms remain.
        rem = eq - recon
        if rem != 0:
            print("Nonlinear or unrecognized leftover term:")
            print(rem)
            raise ValueError("Equation is not linear in the chosen unknowns.")

        rows.append(row)
        rhs.append(K(-const))

    M = Matrix(K, rows)
    b = vector(K, rhs)

    sol = M.solve_right(b)

    return {unknowns[i]: sol[i] for i in range(len(unknowns))}


def reduce_into_basis(target, basis, h_degree=6, h_den=None, label=""):
    """
    Reduce target r(x) dx/y into a chosen basis modulo exacts.

    Finds coefficients c_i and a rational h such that

        target = sum_i c_i basis_i + L_exact(h).

    We use the ansatz

        h = q(x)/h_den,

    with q a polynomial of degree <= h_degree.

    For these GM reductions, h_den = P is usually the right default.
    """
    if h_den is None:
        h_den = P

    target = Fx(target)
    basis = [Fx(b) for b in basis]
    h_den = Fx(h_den)

    n_basis = len(basis)

    # Unknowns for basis coefficients
    Cpoly = PolynomialRing(K, [f'c{i}' for i in range(n_basis)] +
                              [f'a{k}' for k in range(h_degree + 1)])
    gens = Cpoly.gens()

    c_vars = gens[:n_basis]
    a_vars = gens[n_basis:]

    # Need to lift all rational functions into FractionField(Cpoly[x]).
    # Easiest route: build a new polynomial ring in x over Frac(Cpoly).
    KC = Cpoly.fraction_field()
    RX = PolynomialRing(KC, 'x')
    xx = RX.gen()
    FX = FractionField(RX)

    def lift_to_FX(f):
        """
        Lift rational function f(x) over K to rational function over KC.
        """
        f = Fx(f)
        num = R(f.numerator())
        den = R(f.denominator())

        num_lift = RX(0)
        for k in range(num.degree() + 1):
            num_lift += KC(num.monomial_coefficient(x^k)) * xx^k

        den_lift = RX(0)
        for k in range(den.degree() + 1):
            den_lift += KC(den.monomial_coefficient(x^k)) * xx^k

        return FX(num_lift) / FX(den_lift)

    targetX = lift_to_FX(target)
    basisX = [lift_to_FX(b) for b in basis]
    PX = lift_to_FX(P)
    PprimeX = PX.derivative(xx)
    hdenX = lift_to_FX(h_den)

    qX = sum(KC(a_vars[k]) * xx^k for k in range(h_degree + 1))
    hX = FX(qX) / hdenX

    LhX = PX*hX.derivative(xx) + FX(KC(1)/KC(2))*PprimeX*hX

    comboX = sum(KC(c_vars[i]) * basisX[i] for i in range(n_basis))

    exprX = targetX - comboX - LhX

    # Clear denominators.
    numX = RX(exprX.numerator())

    eqs = []
    for k in range(numX.degree() + 1):
        eqs.append(Cpoly(numX.monomial_coefficient(xx^k)))

    unknowns = list(c_vars) + list(a_vars)

    sol = solve_linear_system(eqs, unknowns)

    coeffs = [K(sol[c_vars[i]]) for i in range(n_basis)]

    q = Fx(0)
    for k in range(h_degree + 1):
        q += Fx(K(sol[a_vars[k]])) * x^k

    h = Fx(q) / h_den

    residual = Fx(target - sum(coeffs[i]*basis[i] for i in range(n_basis)) - L_exact(h))


    if residual != 0:
        raise ArithmeticError(f"Nonzero residual in reduction {label}")

    return coeffs, h, residual



# ------------------------------------------------------------
# 7. Compute meromorphic action-frame GM matrices U_lambda
# ------------------------------------------------------------
def action_connection_matrix(lam, h_degree=10):
    """
    Compute U_lam in the tilted meromorphic action frame.

    For the meromorphic action frame we allow the exact primitive h
    to have poles at the finite pole divisor D as well as at P:

        h = q(x)/(D P).

    This is more flexible than h=q/P and is usually needed once
    omega_mu0 and omega_mu1 are included.
    """
    Umat = Matrix(K, n_omega, n_omega)
    exacts = {}

    H_action = Fx(D*P)

    for j, bj in enumerate(omega_basis):
        print(f"  reducing d_{lam} {omega_names[j]} ...", end="", flush=True)

        target = param_deriv_form(bj, lam)

        coeffs, h, residual = reduce_into_basis(
            target,
            omega_basis,
            h_degree=h_degree,
            h_den=H_action,
            label=f"U_{lam}_{omega_names[j]}"
        )

        for i in range(n_omega):
            Umat[i, j] = coeffs[i]

        exacts[j] = h

        print(" done")

    return Umat, exacts


# ------------------------------------------------------------
# Quiet computation of action-frame GM matrices
# ------------------------------------------------------------

#U = {}
#U_exact = {}

#for lam, name in zip(params, param_names):
#    print(f"Computing action-frame connection U_{name} ...", end="", flush=True)
#    U[lam], U_exact[lam] = action_connection_matrix(lam, h_degree=8)
#    print(" done")

# ------------------------------------------------------------
# 8. Embedding S of compact frame into action frame
# ------------------------------------------------------------

# Action-frame ordering:
#
# omega = (omega0, omegaE, omegamu0, omegamu1, omegamu2)^T
#
# eta0 = omegaE
#
# eta1 = omega0 - 2 mu0 omegamu0
#              - 2 mu1 omegamu1
#              -   mu2 omegamu2

S = Matrix(K, n_omega, n_eta, [
    0, 1,
    1, 0,
    0, -2*mu0,
    0, -2*mu1,
    0, -mu2
])

print("\nEmbedding matrix S:")
print(S)


def dS_param(lam):
    """
    Parameter derivative of S.
    """
    return Matrix(K, n_omega, n_eta, [
        S[i, j].derivative(lam)
        for i in range(n_omega)
        for j in range(n_eta)
    ])


# ------------------------------------------------------------
# 9. Fast descended action-frame compact connection
# ------------------------------------------------------------

def S_column(j):
    """
    Return column j of S as a plain list of rational functions.
    """
    return [Fx(S[i, j]) for i in range(n_omega)]


def d_action_combination_to_r(coeffs, lam):
    """
    Differentiate an action-frame combination

        sum_i coeffs[i] omega_i

    directly, without computing the full 5x5 action-frame U matrix.

    Returns the rational coefficient multiplying dx/y.
    """
    total = Fx(0)

    for i in range(n_omega):
        ci = Fx(coeffs[i])

        # derivative of the coefficient
        dci = Fx(ci.derivative(lam))

        # derivative of the basis differential
        dbi = param_deriv_form(omega_basis[i], lam)

        total += dci * omega_basis[i] + ci * dbi

    return Fx(total)


def compact_connection_from_action_fast(lam, h_degree=10):
    """
    Compute B_lam from the embedded action-frame compact basis directly.

    This computes

        d_lam eta_j = d_lam(sum_i S_ij omega_i)

    and then reduces the result into the compact basis.
    """
    Bmat = Matrix(K, n_eta, n_eta)
    exacts = {}

    for j in range(n_eta):
        print(f"  fast reducing descended d_{lam} {eta_names[j]} ...", end="", flush=True)

        coeffs_action = S_column(j)
        target = d_action_combination_to_r(coeffs_action, lam)

        coeffs, h, residual = reduce_into_basis(
            target,
            eta_basis,
            h_degree=h_degree,
            h_den=P,
            label=f"B_from_action_fast_{lam}_{eta_names[j]}"
        )

        for i in range(n_eta):
            Bmat[i, j] = coeffs[i]

        exacts[j] = h

        print(" done")

    return Bmat, exacts

# ------------------------------------------------------------
# Quiet computation of compact matrices from action descent
# ------------------------------------------------------------

B_from_action = {}
B_from_action_exact = {}

for lam, name in zip(params, param_names):
    print(f"\nComputing compact B_{name} from fast action descent ...")
    B_from_action[lam], B_from_action_exact[lam] = (
        compact_connection_from_action_fast(lam, h_degree=10)
    )
    print("done")


# ------------------------------------------------------------
# 10. Direct compact GD-style reduction
# ------------------------------------------------------------

def compact_connection_direct(lam, h_degree=10):
    """
    Direct compact reduction.
    """
    Bmat = Matrix(K, n_eta, n_eta)
    exacts = {}

    H_compact = Fx(P)

    for j, bj in enumerate(eta_basis):
        print(f"  reducing direct d_{lam} {eta_names[j]} ...", end="", flush=True)

        target = param_deriv_form(bj, lam)

        coeffs, h, residual = reduce_into_basis(
            target,
            eta_basis,
            h_degree=h_degree,
            h_den=H_compact,
            label=f"B_direct_{lam}_{eta_names[j]}"
        )

        for i in range(n_eta):
            Bmat[i, j] = coeffs[i]

        exacts[j] = h

        print(" done")

    return Bmat, exacts

# ------------------------------------------------------------
# Quiet computation of direct compact matrices
# ------------------------------------------------------------
B_direct = {}
B_direct_exact = {}

for lam, name in zip(params, param_names):
    print(f"\nComputing direct compact B_{name} ...")
    B_direct[lam], B_direct_exact[lam] = compact_connection_direct(lam, h_degree=10)
    print("done")


# ------------------------------------------------------------
# 11. Harmonization test
# ------------------------------------------------------------

print("\n============================================================")
print("Harmonization checks")
print("============================================================")

Z2 = Matrix(K, n_eta, n_eta, 0)

for lam, name in zip(params, param_names):
    Delta = B_from_action[lam] - B_direct[lam]

    if Delta == Z2:
        print(f"Delta B_{name}: PASS")
    else:
        print(f"Delta B_{name}: FAIL")
        print("Nonzero entries:")
        for i in range(n_eta):
            for j in range(n_eta):
                if Delta[i, j] != 0:
                    print(f"  ({i},{j}) =", Delta[i, j])


# ------------------------------------------------------------
# 12. Optional perturbative substitution mu1=eps nu1, mu2=eps nu2
# ------------------------------------------------------------

def perturbative_expand_matrix(M, order=1):
    """
    Substitute mu1=eps*nu1, mu2=eps*nu2 and expand to requested order.

    This is optional. It is useful after the exact tilted matrices have
    been computed.
    """
    Kpert_base = PolynomialRing(QQ, ['E', 'mu0', 'nu1', 'nu2', 'eps'])
    EE, mmu0, nu1, nu2, eps = Kpert_base.gens()
    Kpert = Kpert_base.fraction_field()

    def conv_expr(a):
        s = str(a)
        # Sage-friendly string substitution.
        s = s.replace('mu1', '(eps*nu1)')
        s = s.replace('mu2', '(eps*nu2)')
        return Kpert(s)

    Mp = Matrix(Kpert, M.nrows(), M.ncols())

    for i in range(M.nrows()):
        for j in range(M.ncols()):
            expr = conv_expr(M[i, j])
            num = expr.numerator()
            den = expr.denominator()

            # Formal expansion by using denominator inverse as series.
            # This simple version is mostly useful when denominators are
            # not too complicated in eps.
            Sseries = expr.taylor(eps, 0, order + 1)
            Mp[i, j] = Sseries

    return Mp
```

```python
# Fast specialized harmonization check at a generic parameter point.
# This avoids the expensive full symbolic elimination while still
# checking the basis-change logic and reduction pipeline.

subs_test = {
    E:   QQ(7)/5,
    mu0: QQ(2)/7,
    mu1: QQ(1)/6,
    mu2: QQ(1)/5,
}


def specialize_fx(f, subs):
    """
    Specialize parameter coefficients of f(x) in Fx = Frac(K[x])
    by substituting values in K = Frac(QQ[E,mu0,mu1,mu2]).
    """
    f = Fx(f)
    num = R(f.numerator())
    den = R(f.denominator())

    num_sp = R(0)
    den_sp = R(0)

    for k in range(num.degree() + 1):
        ck = K(num.monomial_coefficient(x^k)).subs(subs)
        num_sp += R(K(ck)) * x^k

    for k in range(den.degree() + 1):
        ck = K(den.monomial_coefficient(x^k)).subs(subs)
        den_sp += R(K(ck)) * x^k

    return Fx(num_sp) / Fx(den_sp)


def reduce_into_basis_specialized(target, basis, P_local, h_degree=6, h_den=None, label=""):
    """
    Run reduce_into_basis with a specialized curve polynomial P_local by
    temporarily swapping the global P, Px used by L_exact/residual checks.
    """
    global P, Px

    P_old, Px_old = P, Px
    try:
        P = Fx(P_local)
        Px = P.derivative(x)

        if h_den is None:
            h_den = P

        return reduce_into_basis(
            Fx(target),
            [Fx(b) for b in basis],
            h_degree=h_degree,
            h_den=Fx(h_den),
            label=label,
        )
    finally:
        P, Px = P_old, Px_old


def robust_reduce_specialized(target, basis, P_local, label_prefix):
    """
    Try a small sequence of reduction ansaetze to avoid false failures from
    too-restrictive h choices.
    """
    D_local = Fx(D)
    attempts = [
        (6, P_local),
        (8, P_local),
        (10, P_local),
        (8, D_local*P_local),
        (10, D_local*P_local),
    ]

    last_error = None
    for hdeg, hden in attempts:
        try:
            coeffs, h, residual = reduce_into_basis_specialized(
                target,
                basis,
                P_local,
                h_degree=hdeg,
                h_den=hden,
                label=f"{label_prefix}_deg{hdeg}",
            )
            return coeffs, h, residual, hdeg, hden
        except Exception as ex:
            last_error = ex

    raise last_error


def compact_connection_from_action_fast_specialized(lam, subs):
    Bmat = Matrix(K, n_eta, n_eta)

    P_local = specialize_fx(P, subs)
    eta_local = [specialize_fx(b, subs) for b in eta_basis]

    for j in range(n_eta):
        target = Fx(0)
        for i in range(n_omega):
            ci_sym = K(S[i, j])
            ci = Fx(ci_sym.subs(subs))
            dci = Fx(K(ci_sym.derivative(lam)).subs(subs))

            dbi = specialize_fx(param_deriv_form(omega_basis[i], lam), subs)
            obi = specialize_fx(omega_basis[i], subs)
            target += dci*obi + ci*dbi

        coeffs, _, _, used_deg, used_hden = robust_reduce_specialized(
            target,
            eta_local,
            P_local,
            label_prefix=f"B_from_action_sp_{lam}_{eta_names[j]}",
        )

        for i in range(n_eta):
            Bmat[i, j] = coeffs[i]

    return Bmat


def compact_connection_direct_specialized(lam, subs):
    Bmat = Matrix(K, n_eta, n_eta)

    P_local = specialize_fx(P, subs)
    eta_local = [specialize_fx(b, subs) for b in eta_basis]

    for j in range(n_eta):
        target = specialize_fx(param_deriv_form(eta_basis[j], lam), subs)

        coeffs, _, _, used_deg, used_hden = robust_reduce_specialized(
            target,
            eta_local,
            P_local,
            label_prefix=f"B_direct_sp_{lam}_{eta_names[j]}",
        )

        for i in range(n_eta):
            Bmat[i, j] = coeffs[i]

    return Bmat


print("Specialized test point:", subs_test)
print("adaptive h_degree/h_den attempts enabled")
print("\n============================================================")
print("Specialized harmonization checks")
print("============================================================")

Z2 = Matrix(K, n_eta, n_eta, 0)

for lam, name in zip(params, param_names):
    print(f"Checking lambda = {name} ...", end=" ", flush=True)

    B_sp_action = compact_connection_from_action_fast_specialized(lam, subs_test)
    B_sp_direct = compact_connection_direct_specialized(lam, subs_test)

    Delta = B_sp_action - B_sp_direct

    if Delta == Z2:
        print("PASS")
    else:
        print("FAIL")
        print("Delta matrix:")
        print(Delta)
```

```python
# Export B_lambda matrices with defining data and conventions.

import json
from datetime import datetime, timezone

print("============================================================")
print("Exporting B_lambda matrices and definitions")
print("============================================================")

# If matrix dictionaries are not already present, compute a lightweight version.
if 'B_from_action' not in globals() or 'B_direct' not in globals():
    print("B matrices not found in memory; computing with h_degree=6 ...")
    B_from_action = {}
    B_direct = {}
    for lam in params:
        B_from_action[lam], _ = compact_connection_from_action_fast(lam, h_degree=6)
        B_direct[lam], _ = compact_connection_direct(lam, h_degree=6)

# Conventions/definitions to export alongside matrices.
definitions = {
    "connection_convention": "partial_lambda(eta) = B_lambda * eta (column-image convention)",
    "parameters": ["E", "mu0", "mu1", "mu2"],
    "curve": "y^2 = P(x), P(x) = -2*(x^4 - mu2*x^3 + (E-2)*x^2 + (mu1+mu2)*x + mu0 - E + 1)",
    "finite_pole_divisor": "D(x) = 1 - x^2",
    "omega_basis_order": ["omega0 = P/D * dx/y", "omegaE = dx/y", "omegamu0 = -1/D * dx/y", "omegamu1 = -x/D * dx/y", "omegamu2 = -x * dx/y"],
    "eta_basis": ["eta0 = omegaE = dx/y", "eta1 = omega0 - 2*mu0*omegamu0 - 2*mu1*omegamu1 - mu2*omegamu2 = (2*(x^2+E-1)-mu2*x) * dx/y"],
    "embedding_S_rows_in_omega_order": [
        ["0", "1"],
        ["1", "0"],
        ["0", "-2*mu0"],
        ["0", "-2*mu1"],
        ["0", "-mu2"],
    ],
    "descent_relation": "(partial_lambda S + U_lambda S) = S B_lambda mod exact forms",
}

lam_keys = [(E, "E"), (mu0, "mu0"), (mu1, "mu1"), (mu2, "mu2")]
zero_mat = Matrix(K, n_eta, n_eta, 0)


def entry_clean(expr):
    """Return a human-readable factored form of a rational function entry."""
    return latex_factor0(Fx(expr))


def matrix_clean_strings(M):
    return [[entry_clean(M[i, j]) for j in range(M.ncols())] for i in range(M.nrows())]


def matrix_raw_strings(M):
    return [[str(M[i, j]) for j in range(M.ncols())] for i in range(M.nrows())]


def latex_factored_matrix(M):
    rows = []
    for i in range(M.nrows()):
        row = []
        for j in range(M.ncols()):
            row.append(entry_clean(M[i, j]))
        rows.append(" & ".join(row))
    return r"\begin{pmatrix}" + r"\\".join(rows) + r"\end{pmatrix}"

export_obj = {
    "timestamp_utc": datetime.now(timezone.utc).isoformat(),
    "definitions": definitions,
    "matrices": {}
}

for lam, name in lam_keys:
    Bfa = B_from_action[lam]
    Bdi = B_direct[lam]
    Delta = Bfa - Bdi

    export_obj["matrices"][name] = {
        "B_from_action": matrix_raw_strings(Bfa),
        "B_direct": matrix_raw_strings(Bdi),
        "Delta": matrix_raw_strings(Delta),
        "B_from_action_factored": matrix_clean_strings(Bfa),
        "B_direct_factored": matrix_clean_strings(Bdi),
        "Delta_factored": matrix_clean_strings(Delta),
        "delta_is_zero": bool(Delta == zero_mat),
    }

base_name = "B_lambda_export"
json_path = f"{base_name}.json"
txt_path = f"{base_name}.txt"
tex_path = f"{base_name}.tex"

# JSON export
with open(json_path, "w", encoding="utf-8") as f:
    json.dump(export_obj, f, indent=2)

# Plain text export
lines = []
lines.append("B_lambda Export")
lines.append("=" * 60)
lines.append(f"Generated UTC: {export_obj['timestamp_utc']}")
lines.append("")
lines.append("Definitions")
lines.append("-" * 60)
for k, v in definitions.items():
    lines.append(f"{k}: {v}")
lines.append("")

for _, name in lam_keys:
    lines.append(f"lambda = {name}")
    lines.append("-" * 60)
    lines.append("B_from_action (raw):")
    for row in export_obj["matrices"][name]["B_from_action"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append("B_from_action (factored):")
    for row in export_obj["matrices"][name]["B_from_action_factored"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append("B_direct (raw):")
    for row in export_obj["matrices"][name]["B_direct"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append("B_direct (factored):")
    for row in export_obj["matrices"][name]["B_direct_factored"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append("Delta = B_from_action - B_direct (raw):")
    for row in export_obj["matrices"][name]["Delta"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append("Delta (factored):")
    for row in export_obj["matrices"][name]["Delta_factored"]:
        lines.append("  [" + ", ".join(row) + "]")
    lines.append(f"delta_is_zero: {export_obj['matrices'][name]['delta_is_zero']}")
    lines.append("")

with open(txt_path, "w", encoding="utf-8") as f:
    f.write("\n".join(lines))

# TeX-ish export (easy to paste into notes)
tex_lines = []
tex_lines.append("% Auto-generated B_lambda export")
tex_lines.append("\\section*{B_\\lambda Export}")
tex_lines.append("\\subsection*{Definitions}")
tex_lines.append("\\begin{itemize}")
for k, v in definitions.items():
    tex_lines.append(f"  \\item {k}: {v}")
tex_lines.append("\\end{itemize}")

for lam, name in lam_keys:
    Bfa = B_from_action[lam]
    Bdi = B_direct[lam]
    Delta = Bfa - Bdi

    tex_lines.append(f"\\subsection*{{$\\lambda={name}$}}")
    tex_lines.append("\\paragraph{Raw matrices}")
    tex_lines.append("$B^{\\mathrm{from\\ action}}_\\lambda =$")
    tex_lines.append(latex(Bfa))
    tex_lines.append("\\\n$B^{\\mathrm{direct}}_\\lambda =$")
    tex_lines.append(latex(Bdi))
    tex_lines.append("\\\n$\\Delta B_\\lambda = B^{\\mathrm{from\\ action}}_\\lambda - B^{\\mathrm{direct}}_\\lambda =$")
    tex_lines.append(latex(Delta))

    tex_lines.append("\\paragraph{Factored presentation}")
    tex_lines.append("$B^{\\mathrm{from\\ action}}_\\lambda$ factored:")
    tex_lines.append("$$" + latex_factored_matrix(Bfa) + "$$")
    tex_lines.append("$B^{\\mathrm{direct}}_\\lambda$ factored:")
    tex_lines.append("$$" + latex_factored_matrix(Bdi) + "$$")
    tex_lines.append("$\\Delta B_\\lambda$ factored:")
    tex_lines.append("$$" + latex_factored_matrix(Delta) + "$$")
    tex_lines.append(f"\\\nZero check: {Delta == zero_mat}")

with open(tex_path, "w", encoding="utf-8") as f:
    f.write("\n\n".join(tex_lines))

print(f"Wrote: {json_path}")
print(f"Wrote: {txt_path}")
print(f"Wrote: {tex_path}")
print("Done.")
```
